Cogemos algunos de los datos que parecen más interesantes desde la API de OpenSky, haciendo una llamada de prueba

Nota: Hace falta instalar el paquete de opensky_api: pip install "git+https://github.com/openskynetwork/opensky-api.git#subdirectory=python"

```python
from opensky_api import OpenSkyApi, TokenManager

#SOlo hace falta registrarse para tener x10 más tokens para llamadas (no es estrictamente necesario); borrar lo de token_manager si no quereis
api = OpenSkyApi(token_manager=TokenManager.from_json_file("credentials.json"))
```

```python
states = api.get_states()
departures = api.get_departures_by_airport("EDDF", begin=1517227200, end=1517230800)
arrivals = api.get_arrivals_by_airport("EDDF", begin=1517227200, end=1517230800)
flights = api.get_flights_from_interval(begin=1517227200, end=1517230800)
```

Paso los objetos a dataframe de Pandas para más fácil manipulación

```python
states_df = pd.DataFrame([flight.__dict__ for flight in states.states])
departures_df = pd.DataFrame([flight.__dict__ for flight in departures])
arrivals_df = pd.DataFrame([flight.__dict__ for flight in arrivals])
flights_df = pd.DataFrame([flight.__dict__ for flight in flights])
```
